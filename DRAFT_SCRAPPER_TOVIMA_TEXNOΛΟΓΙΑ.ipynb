```python
article_urls = set()

for i in range(1, 7):

    if i == 1:
        url = "https://www.tovima.gr/category/texnologia/"
    else:
        url = f"https://www.tovima.gr/category/texnologia/page/{i}/"

    print("Scraping:", url)

    response = requests.get(url, headers=headers)
    doc = BeautifulSoup(response.text, "html.parser")

    articles = doc.find_all(
        attrs={"itemprop": "itemListElement"}
    )

    for article in articles:

        url_meta = article.find(
            "meta",
            {"itemprop": "url"}
        )

        if url_meta:

            article_url = url_meta.get("content")

            if article_url and "/2026/" in article_url:
                article_urls.add(article_url)
```
